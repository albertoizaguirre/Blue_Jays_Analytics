
# Toronto Blue Jays: Final Analysis & Conclusions (2025 vs. 2026)

## 1. Executive Summary

### Research Question

**What statistical factors contributed to the Toronto Blue Jays' decline between the 2025 and 2026 MLB seasons?**

### Project Overview

The Toronto Blue Jays declined from a 94–68 record and first-place finish in the American League East in 2025 to a 79–83 record and fifth-place finish in 2026.

This project investigated the decline through four analytical areas:

- Offensive production
- Pitching and run prevention
- Defensive performance
- Roster composition and player participation

Using Python, pandas, NumPy, and matplotlib, cleaned team-level and player-level datasets were analyzed to identify meaningful statistical changes between the two seasons.

### Key Findings

**Offensive Decline**

Toronto scored 150 fewer runs in 2026, while home runs declined by 39 and team OPS fell from .761 to .692. Most qualified returning hitters experienced declining individual production.

**Improved Run Prevention**

Despite weaker performances among several returning pitchers, Toronto allowed 27 fewer runs in 2026. Team ERA improved from 4.19 to 3.89, supported by stronger collective results among 2026-only pitchers.

**Mixed Defensive Results**

Traditional fielding statistics deteriorated, but advanced defensive metrics improved. Defensive Runs Saved increased from +4 to +33, suggesting better performance in aspects of defense not fully captured by errors and fielding percentage.

**Roster Changes and Player Availability**

Only 25 of Toronto's 58 players from 2025 also appeared in 2026. Injuries, reduced appearances, personnel changes, and declining performances among established hitters provided additional context for the offensive decline.

### Primary Conclusion

**Declining offensive production was the primary statistical contributor to Toronto's worsening run differential.**

The team's run differential fell from +77 in 2025 to -46 in 2026, a decline of 123 runs.

Improved run prevention partially offset the offensive decline, but it was insufficient to prevent the team's overall deterioration.

These findings identify statistical relationships rather than establishing definitive causal explanations.



## 2. Team Performance Overview

To summarize Toronto's overall decline, key team performance indicators are compared between the 2025 and 2026 seasons.

The comparison focuses on wins, losses, offensive production, run prevention, and run differential.

These measures establish the overall magnitude of Toronto's decline and provide context for the individual analytical findings.


```python

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Consolidate key findings from previous notebooks
team_summary = pd.DataFrame({
    "Statistic": [
        "Wins",
        "Losses",
        "Runs Scored",
        "Runs Allowed",
        "Run Differential",
        "Home Runs",
        "OPS",
        "ERA"
    ],
    "2025": [
        94, 68, 798, 721, 77, 191, 0.761, 4.19
    ],
    "2026": [
        79, 83, 648, 694, -46, 152, 0.692, 3.89
    ]
})

# Calculate the change between seasons
team_summary["Change"] = (
    team_summary["2026"] - team_summary["2025"]
)

# Display the summary
display(team_summary.set_index("Statistic"))
```


## 3. Explaining the Run Differential Decline

Toronto's run differential declined from +77 in 2025 to -46 in 2026, representing a deterioration of 123 runs.

To understand this change, the differences in runs scored and runs allowed are examined together.

The team scored 150 fewer runs in 2026 but also allowed 27 fewer runs. Consequently, improved run prevention partially offset the substantial decline in offensive production.

The following visualization illustrates how these two changes contributed to the overall deterioration in run differential.


```python

# Values from our team performance summary
rd_2025 = 77
offensive_change = -150
run_prevention_change = 27
rd_2026 = -46

# Calculate each stage of the run differential
after_offense = rd_2025 + offensive_change

# Create the waterfall chart
fig, ax = plt.subplots(figsize=(10, 6))

labels = [
    "2025 Run\nDifferential",
    "Offensive\nDecline",
    "Improved Run\nPrevention",
    "2026 Run\nDifferential"
]

# Starting and ending columns
ax.bar(0, rd_2025, color="#4472C4", width=0.6)
ax.bar(3, rd_2026, color="#4472C4", width=0.6)

# Offensive decline: from +77 to -73
ax.bar(
    1, offensive_change,
    bottom=rd_2025,
    color="#C0504D",
    width=0.6
)

# Improved run prevention: from -73 to -46
ax.bar(
    2, run_prevention_change,
    bottom=after_offense,
    color="#70AD47",
    width=0.6
)

# Connect each stage
ax.plot([0.3, 0.7], [rd_2025, rd_2025],
        color="gray", linestyle="--")

ax.plot([1.3, 1.7], [after_offense, after_offense],
        color="gray", linestyle="--")

ax.plot([2.3, 2.7], [rd_2026, rd_2026],
        color="gray", linestyle="--")

# Add labels to each bar
ax.text(0, 82, "+77", ha="center", fontweight="bold")

# Place the offensive decline label inside the red bar
ax.text(
    1, 0, "-150",
    ha="center",
    va="center",
    color="white",
    fontweight="bold",
    fontsize=12
)

# Place the run prevention improvement label above the green bar
ax.text(2, -41, "+27", ha="center", fontweight="bold")

# Place the final run differential label below the final bar
ax.text(3, -55, "-46", ha="center", fontweight="bold")

# Format chart
ax.axhline(0, color="black", linewidth=0.8)
ax.set_xticks(range(4))
ax.set_xticklabels(labels)
ax.set_ylabel("Run Differential")
ax.set_title(
    "Toronto Blue Jays: Explaining the Run Differential Decline\n2025 vs. 2026"
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```


## 4. Key Analytical Findings

### 4.1 Offensive Performance — Primary Area of Decline

Toronto's offensive production deteriorated substantially between 2025 and 2026.

- Runs scored declined from 798 to 648 (-150).
- Home runs decreased from 191 to 152 (-39).
- Team OPS fell from .761 to .692.
- Eight of nine qualified returning hitters recorded lower OPS in 2026.

The decline among returning hitters was particularly significant. Although 2026-only players provided more collective offensive production than the 2025-only group, their contributions did not offset the deterioration among established hitters.

**Finding:** Declining offensive production, particularly among returning players, was the largest statistical contributor to Toronto's worsening run differential.

### 4.2 Pitching Performance — Improved Overall Run Prevention

Toronto's overall pitching statistics improved despite weaker performances among several returning pitchers.

- Team ERA decreased from 4.19 to 3.89.
- Runs allowed declined from 721 to 694 (-27).
- Seven of ten qualified returning pitchers recorded higher ERAs.
- The 2026-only pitching group recorded a collective ERA of 3.64, compared with 4.68 among 2025-only pitchers.

These findings suggest that changes in pitching personnel helped offset declining performances among returning pitchers.

**Finding:** Overall run prevention improved, meaning pitching was not the primary statistical driver of Toronto's worsening run differential.

### 4.3 Defensive Performance — Mixed Traditional and Advanced Results

Toronto's traditional defensive statistics deteriorated, while advanced defensive metrics improved.

- Errors increased from 86 to 99.
- Fielding percentage declined from .985 to .983.
- Defensive Runs Saved increased from +4 to +33.
- Total Zone Runs improved from -8 to +8.

The advanced metrics suggest improved defensive performance in aspects not fully captured by traditional fielding statistics.

**Finding:** Improved advanced defensive metrics may have contributed to Toronto's stronger overall run prevention, although their independent impact cannot be established.

### 4.4 Roster Composition — Turnover and Reduced Availability

Toronto experienced substantial roster turnover and changes in player participation.

- Only 25 of the 58 players who appeared in 2025 returned in 2026.
- Thirty-nine players appeared for Toronto in 2026 but not in 2025.
- Addison Barger and Alejandro Kirk experienced major reductions in appearances due to injuries.
- Other changes reflected trades, performance-related assignments, and expanded opportunities for emerging players.

These developments provide additional context for the decline among returning hitters and the improved contributions from 2026-only pitchers.

**Finding:** Roster changes, reduced availability, and declining performances among established contributors were associated with Toronto's overall decline, particularly on offense.



## 5. Final Conclusion

### What Went Wrong with the Toronto Blue Jays?

The Toronto Blue Jays' decline from 94 wins in 2025 to 79 wins in 2026 was primarily associated with a substantial deterioration in offensive production rather than worsening overall run prevention.

Toronto scored 150 fewer runs, hit 39 fewer home runs, and experienced a decline in team OPS from .761 to .692. Eight of nine qualified returning hitters recorded lower OPS, demonstrating that reduced production among established players was a significant component of the team's offensive struggles.

Roster changes and reduced player availability provided additional context. Injuries affected important contributors such as Addison Barger and Alejandro Kirk, while the departure of Bo Bichette and declining performances among returning hitters further characterized the changing offense.

In contrast, Toronto's overall run prevention improved. Runs allowed decreased by 27, team ERA fell from 4.19 to 3.89, and advanced defensive metrics indicated improvement despite an increase in recorded fielding errors. Contributions from the 2026-only pitching group also helped offset weaker performances among returning pitchers.

Ultimately, Toronto's 150-run offensive decline outweighed its 27-run improvement in run prevention, causing the team's run differential to deteriorate from +77 to -46.

### Final Assessment

**The primary statistical explanation for Toronto's decline was reduced offensive production, particularly among returning hitters, compounded by changes in roster composition and player availability.**

Although pitching and advanced defensive performance improved, those gains were insufficient to compensate for the decline in scoring.

While these findings do not establish definitive causal relationships, they provide a clear, evidence-based explanation of the statistical factors associated with Toronto's disappointing 2026 season.
