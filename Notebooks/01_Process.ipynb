# Toronto Blue Jays 2025 vs. 2026 : Data Cleaning

## Process Phase

### Project Objective

This case study investigates the changes in the Toronto Blue Jays' performance between the 2025 and 2026 MLB seasons.

The purpose of the Process phase is to prepare the collected datasets for analysis by removing unnecessary information, standardizing player names, handling missing values, and ensuring consistency between seasons.

### Data Cleaning Approach

The analysis uses nine datasets covering team batting, pitching, fielding, individual player statistics, and team rosters.

The original CSV files are preserved in the `data/raw` folder, while processed versions are saved separately in `data/processed`.

```python
# Import pandas for data manipulation
import pandas as pd

# Load player batting data for both seasons
player_batting_2025 = pd.read_csv("../data/raw/player_batting_2025.csv")
player_batting_2026 = pd.read_csv("../data/raw/player_batting_2026.csv")

# Identify columns that are not needed for our analysis
columns_to_remove = [
    "Rk",
    "Age",
    "Pos.1",
    "Awards",
    "Player-additional"
]

# Remove unnecessary columns from both datasets
player_batting_2025 = player_batting_2025.drop(columns=columns_to_remove)
player_batting_2026 = player_batting_2026.drop(columns=columns_to_remove)
```

```python
# Remove team totals to avoid double-counting player statistics
player_batting_2025 = player_batting_2025[
    player_batting_2025["Player"] != "Team Totals"
]

player_batting_2026 = player_batting_2026[
    player_batting_2026["Player"] != "Team Totals"
]
```

```python
# Remove batting-handedness symbols from player names
player_batting_2025["Player"] = (
    player_batting_2025["Player"].str.replace(r"[*#]", "", regex=True)
)

player_batting_2026["Player"] = (
    player_batting_2026["Player"].str.replace(r"[*#]", "", regex=True)
)
```

```python
# Check for duplicate player names in both seasons
print("2025 duplicates:", player_batting_2025["Player"].duplicated().sum())

print("2026 duplicates:", player_batting_2026["Player"].duplicated().sum())
```

```python
# Save cleaned player batting datasets
player_batting_2025.to_csv(
    "../data/processed/player_batting_2025_clean.csv",
    index=False
)

player_batting_2026.to_csv(
    "../data/processed/player_batting_2026_clean.csv",
    index=False
)
```

```python
# Load player pitching data for both seasons
player_pitching_2025 = pd.read_csv("../data/raw/player_pitching_2025.csv")
player_pitching_2026 = pd.read_csv("../data/raw/player_pitching_2026.csv")
```

```python
# Remove unnecessary columns
columns_to_remove = ["Awards", "Player-additional"]

player_pitching_2025 = player_pitching_2025.drop(columns=columns_to_remove)
player_pitching_2026 = player_pitching_2026.drop(columns=columns_to_remove)

# Remove Team Totals rows
player_pitching_2025 = player_pitching_2025[
    player_pitching_2025["Player"] != "Team Totals"
].copy()

player_pitching_2026 = player_pitching_2026[
    player_pitching_2026["Player"] != "Team Totals"
].copy()

# Remove asterisks from left-handed pitchers' names
player_pitching_2025["Player"] = (
    player_pitching_2025["Player"].str.replace("*", "", regex=False)
)

player_pitching_2026["Player"] = (
    player_pitching_2026["Player"].str.replace("*", "", regex=False)
)
```

```python
# Save cleaned player pitching datasets
player_pitching_2025.to_csv(
    "../data/processed/player_pitching_2025_clean.csv",
    index=False
)

player_pitching_2026.to_csv(
    "../data/processed/player_pitching_2026_clean.csv",
    index=False
)
```

```python
# Load 2025 and 2026 roster datasets
roster_2025 = pd.read_csv("../data/raw/roster_2025.csv")
roster_2026 = pd.read_csv("../data/raw/roster_2026.csv")

# Keep only player names and games played
roster_2025 = roster_2025[["Player", "G"]].copy()
roster_2026 = roster_2026[["Player", "G"]].copy()

# Remove extra spaces around player names
roster_2025["Player"] = roster_2025["Player"].str.strip()
roster_2026["Player"] = roster_2026["Player"].str.strip()

roster_2025.to_csv(
    "../data/processed/roster_2025_clean.csv",
    index=False
)

roster_2026.to_csv(
    "../data/processed/roster_2026_clean.csv",
    index=False
)
```
